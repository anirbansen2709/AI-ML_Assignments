### Dataset Generation Rules & Ranges (Group 28)

* **Patient ID:** Sequences exactly from `0` to `999` (Total 1000 patients).
* **Number of Medicines ($K$):** The total available treatment choices are calculated as $K = (G \bmod 3) + 5$.
    * For Group 28: $K = 6$ medicines.
* **Hidden Success Probability ($P_i$):** The true probability of successful recovery for a given medicine $i$ is $P_i = 0.4 + ((G+i) \bmod 6) \times 0.07$.
    * *Treatment 0:* 0.6800
    * *Treatment 1:* 0.7500
    * *Treatment 2:* 0.4000
    * *Treatment 3:* 0.4700
    * *Treatment 4:* 0.5400
    * *Treatment 5:* 0.6100
* **Patient Severity:** The inherent disease severity score is defined as $\text{Severity} = (\text{patient\_id} \bmod 5) + 1$.
    * Values range from **1 (mild)** to **5 (critical)**.
* **Clinical Outcome:** A binary value indicating recovery:
    * **1 (Recovered):** Occurs with probability $P_i$.
    * **0 (Not Recovered):** Occurs with probability $1 - P_i$.
* **Utility Logic:** To simulate real-world hospital impact, the treatment benefit scales down for highly severe patients. The customized reward is calculated as:
    $$\text{Utility Score} = \text{clinical\_outcome} \times \left(1 - \frac{\text{Severity}}{10}\right)$$

```python
# ==========================================
#  Task 1: Dataset Design
# ==========================================
class ClinicalTrialEnvironment:
    """
    Encapsulates the MAB environment, isolating the dataset and reward logic.
    """
    def __init__(self, group_id: int):
        """
        Initializes the environment with a group ID.
        """
        self.group_id = group_id
        # Calculate available treatments based on formula K = (G mod 3) + 5
        self.total_medicines = (self.group_id % 3) + 5
        # The arms are the available medical treatments.
        # Enforce reproducibility
        random.seed(self.group_id)
        np.random.seed(self.group_id)
        # Hidden probabilities based on assignment formula: 0.4 + ((G+i) mod 6) * 0.07
        self.true_efficacies = [0.4 + ((self.group_id + i) % 6) * 0.07 for i in range(self.total_medicines)]
        # Initialize the patient dataset
        self.patient_cohort = self._bootstrap_cohort()

    def _bootstrap_cohort(self) -> pd.DataFrame:
        """Bootstraps the 1000-patient dataset with pre-calculated severity scores."""
        records = [{"patient_id": pid, "severity_score": (pid % 5) + 1} for pid in range(1000)]
        return pd.DataFrame(records)

    def display_trial_parameters(self) -> None:
        """Outputs environment metrics in a clean, readable format."""
        print(f"\n--- TRIAL INITIALIZATION (Group {self.group_id}) ---")
        print(f"Available Treatment Arms (K) : {self.total_medicines}")
        print("True Underlying Efficacies   :")
        for idx, eff in enumerate(self.true_efficacies):
            print(f"  -> Treatment_{idx:02d} : {eff:.4f}")
        print("\nPatient Cohort Preview (Top 10):")
        print(self.patient_cohort.head(10).to_string(index=False))
        print("-" * 60)

    def prescribe(self, patient_idx: int, treatment_idx: int) -> Tuple[int, float]:
        """
        Calculates the binary outcome and continuous utility of a treatment.
        Returns: Tuple of (clinical_outcome, utility_score)
        """
        severity = self.patient_cohort.at[patient_idx, 'severity_score']
        efficacy = self.true_efficacies[treatment_idx]
        # This represents whether the patient was successfully treated or not (1 for Success, 0 for Failure)
        # It is determined by pulling a random number against the drug's true hidden efficacy
        outcome = 1 if np.random.random() < efficacy else 0
        # In a real-world scenario, a successful treatment might be less "valuable" or more taxing if the
        # patient has a highly severe condition, or perhaps severe patients have a different utility
        # threshold. The environment calculates a customized reward score Utility formula:
        # clinical_outcome * (1 - (Severity / 10))
        utility = float(outcome * (1.0 - (severity / 10.0)))
        #because our environment tracks a patient's condition, it introduces a light Contextual Bandit element.
        return outcome, utility
```

```python
# Define group ID and initialize the environment
GROUP_ID = 28
env = ClinicalTrialEnvironment(GROUP_ID)
env.display_trial_parameters()
```

### Core Framework: Strategy & Simulation

#### 1. Polymorphic Strategy Base (`BanditStrategy`)
This base class defines the standard interface and shared memory for all Multi-Armed Bandit (MAB) algorithms evaluated in this trial.
* **State Tracking:** Initializes zero-arrays for `prescription_counts` (how many times a medicine was prescribed) and `estimated_efficacies` (the running average of success rates).
* **`select_arm(current_step)`:** An abstract method that each derived strategy (e.g., Epsilon-Greedy, UCB1) must override to apply its specific exploration/exploitation logic.
* **`update_knowledge(arm, outcome)`:** Incrementally updates the agent's internal belief after a treatment outcome is observed, using the running average formula to adjust the estimated efficacy of the prescribed medicine.

#### 2. Experiment Runner (`TrialSimulation`)
This class binds a specific strategy to the `ClinicalTrialEnvironment` and manages the continuous learning loop across the entire patient cohort.
* **Execution Loop:** Iterates through all 1000 patients performing four key steps: **Select** (agent picks medicine) $\rightarrow$ **Resolve** (environment returns binary outcome and utility) $\rightarrow$ **Learn** (agent updates its knowledge) $\rightarrow$ **Log** (record metrics).
* **Data Tracking:** Dynamically populates the patient DataFrame with the `assigned_medicine`, `clinical_outcome`, and `utility_score` as the trial progresses.
* **Verbose Logging:** Fulfills the assignment requirement by printing detailed diagnostic logs for the initial patient and every subsequent 100th patient to monitor step utility and cumulative performance.

```python
# ==========================================
# POLYMORPHIC STRATEGY CLASSES
# ==========================================
class BanditStrategy:
    """Base class defining the interface for all MAB strategies."""
    def __init__(self, n_arms: int, name: str):
        """Instantiates the shared MAB skeleton by initializing zero-arrays for treatment counters and
        # estimated average success rates."""
        self.n_arms = n_arms
        self.name = name
        # how many times each treatment has been prescribed so far.
        self.prescription_counts = np.zeros(n_arms)
        # agent's current belief or running average success rate for each drug.
        self.estimated_efficacies = np.zeros(n_arms)

    def select_arm(self, current_step: int) -> int:
        """Abstract method overridden by specific strategies."""
        raise NotImplementedError("Subclasses must implement arm selection.")

    def update_knowledge(self, arm: int, outcome: int) -> None:
        """ After a treatment is prescribed and the environment returns a binary outcome (1 for success, 0 for
        # failure), this method updates the agent's memory incrementally."""
        self.prescription_counts[arm] += 1
        n = self.prescription_counts[arm]
        # New = Old + (Target - Old) / N
        self.estimated_efficacies[arm] += (outcome - self.estimated_efficacies[arm]) / n

# ==========================================
# EXPERIMENT RUNNER
# ==========================================
class TrialSimulation:
    """Runs a strategy against the environment and logs iteration data."""
    def __init__(self, env: ClinicalTrialEnvironment):
        """Binds the strategy runner to the trial environment and extracts the global iteration horizon limit (T = 1000 patients)."""
        self.env = env
        self.num_patients = len(env.patient_cohort)

    def execute(self, strategy: BanditStrategy) -> Tuple[List[float], pd.DataFrame]:
        print(f"\n>>> INITIATING PROTOCOL: {strategy.name} <<<")
        cumulative_utility = 0.0
        utility_history = []
        # Prepare tracking dataframe
        tracker_df = self.env.patient_cohort.copy()
        tracker_df['assigned_medicine'] = -1
        tracker_df['clinical_outcome'] = -1
        tracker_df['utility_score'] = 0.0
        for t in range(self.num_patients):
            # 1. Agent selects treatment
            # It looks at patient t and asks the strategy to choose one of the 6 available treatments.
            chosen_med = strategy.select_arm(t)
            # 2. Environment resolves outcome
            # It takes that chosen medicine and runs it through the environment. The environment simulates
            # the clinical trial and hands back the binary outcome and the scalar utility
            outcome, utility = self.env.prescribe(t, chosen_med)
            # 3. Agent learns
            # It hands the binary outcome back to the strategy. This is crucial because it allows
            # the algorithm to learn from its success or failure before the next patient arrives.
            strategy.update_knowledge(chosen_med, outcome)
            # 4. Update metrics
            cumulative_utility += utility
            utility_history.append(cumulative_utility)
            # 5. Log to dataframe dynamically
            tracker_df.loc[t, ['assigned_medicine', 'clinical_outcome', 'utility_score']] = [chosen_med, outcome, utility]
            # 6. REQUIRED VERBOSE LOGGING - Prints outputs for EVERY 100th iteration
            if t == 0 or (t + 1) % 100 == 0 or t == self.num_patients - 1:
                print(f"[PATIENT-{t:03d}]  | Protocol: {strategy.name[:10]:<10}  | Med_{chosen_med:02d}  | "
                      f"Outcome: {outcome}  | Step Utility: {utility:.2f}  | Total Utility: {cumulative_utility:.2f}")
        return utility_history, tracker_df
```

## Task 2: Immediate Exploitation Strategy

A hospital administrator suggests: "Once a treatment appears best, continue prescribing only
that treatment for all future patients."

• Implement a recommendation model following this policy.

• Perform 1000 iterations and evaluate cumulative reward.

Hint:
Test each medicine exactly 10 times initially, then continue using only the currently bestperforming one.

```python
# ==========================================
#  Task 2: Immediate Exploitation Strategy
# ==========================================
class ImmediateExploiter(BanditStrategy):
    """Explores round-robin exactly 10 times per arm, then purely exploits."""
    def __init__(self, n_arms: int):
        """Initialize a naive baseline agent that cleanly splits exploration and exploitation."""
        super().__init__(n_arms, "Immediate Exploitation")

    def select_arm(self, current_step: int) -> int:
        # Phase 1: Pure Exploration (The First 60 Patients)
        if current_step < 10 * self.n_arms:
            return current_step % self.n_arms # Round-robin testing
        # Phase 2: Pure Exploitation (Patients 61 to 1000)
        return int(np.argmax(self.estimated_efficacies)) # Pure exploitation
```

### Strategy Analysis: Immediate Exploitation

**Overview:**
This approach divides the clinical trial into two rigid, non-overlapping phases: a strict exploration phase followed by a permanent exploitation phase. By testing each of the 6 medicines exactly 10 times, the agent spends the first 60 patients building a baseline efficacy estimate. For the remaining 940 patients, it greedily prescribes the single medicine that performed best during the initial trials.

**Expected Behavior & Trade-offs:**
* **Rapid Convergence:** The model will lock into its final decision extremely early (at patient 61), making it the fastest strategy to "settle" on a treatment.
* **Zero Long-term Regret (If Correct):** If the best medicine is accurately identified during the first 60 steps, it maximizes utility for all remaining patients.
* **Risk of Premature Lock-in:** The fatal flaw of this naive approach is its vulnerability to variance. With only 10 trials per medicine, a truly superior drug might get "unlucky" and fail early, while a mediocre drug might get "lucky" and succeed. Because the algorithm stops exploring entirely after step 60, it can never correct this initial misjudgment. This often results in a sub-optimal cumulative reward compared to dynamic strategies.

### Task 3: Controlled Clinical Trial Strategy

Doctors suggest: "Most patients should receive the current best treatment, but occasionally
another treatment should be tested to discover hidden opportunities."

• Implement this strategy with: 10% intentional exploration.

• Run the simulation for 1000 patients and evaluate cumulative reward.

• Analyze what happens if exploration changes to: 1%, 50%.

```python
# ===========================================
#  Task 3: Controlled Clinical Trial Strategy
# ===========================================
class EpsilonGreedyStrategy(BanditStrategy):
    """Explores randomly with probability epsilon, otherwise exploits."""
    def __init__(self, n_arms: int, epsilon: float):
        """Saves the constant epsilon factor parameter as the random exploration budget."""
        super().__init__(n_arms, f"Epsilon-Greedy (ε={epsilon})")
        self.epsilon = epsilon #exploration budget

    def select_arm(self, current_step: int) -> int:
        if np.random.random() < self.epsilon:
            # Picks a random treatment from all K options evenly
            return np.random.randint(0, self.n_arms) # Explore
        # Picks treatment with highest estimated efficacy (argmax)
        return int(np.argmax(self.estimated_efficacies)) # Exploit
```

### Strategy Analysis: Controlled Clinical Trial (Epsilon-Greedy)

**Overview:**
The Epsilon-Greedy algorithm provides a continuous, controlled balance between exploration and exploitation. Instead of front-loading all experimentation, it explores randomly with a fixed probability (**ε**) and exploits the best-known treatment with a probability of **1 - ε**. This ensures the system never completely stops learning, allowing it to eventually correct early mistakes.

**Expected Behavior & Impact of Changing Epsilon (ε):**
* **ε = 0.10 (10% Exploration - The Baseline):** This generally hits the "sweet spot" for a 1000-patient trial. It dedicates enough trials to accurately map the hidden success probabilities while exploiting the best medicine 90% of the time to steadily maximize patient recovery and cumulative utility.
* **ε = 0.01 (1% Exploration - Under-exploration):** With only 1 in 100 patients receiving a random treatment, the model learns far too slowly. If its initial random pulls happen to yield poor results for the true "best" medicine, it will take hundreds of iterations to realize its mistake, temporarily locking into a sub-optimal choice.
* **ε = 0.50 (50% Exploration - Over-exploration):** This acts almost like a coin toss between learning and applying knowledge. Even after the algorithm is statistically certain of the optimal treatment, it will still prescribe sub-optimal medicines to half of the patients. In a real-world clinical setting, this is highly unethical and results in a severely degraded cumulative reward.

### Task 4: Confidence-Based Strategy

A senior physician recommends: "Treatments with fewer observations should initially be given
more chances, but this preference should reduce as evidence grows."

• Implement the UCB1 algorithm and evaluate cumulative reward.

```python
# ===================================
#  Task 4: Confidence-Based Strategy
# ===================================

# When an arm has been pulled many times, we are highly confident about its true success rate.
# When an arm has rarely been pulled, its true value is highly uncertain. Instead of just looking
# at the estimated average performance, UCB1 calculates a maximum plausible performance for each drug.
# It chooses whichever medicine has the highest Upper Confidence Bound

class UCB1Strategy(BanditStrategy):
    """Upper Confidence Bound strategy balancing exploration/exploitation mathematically."""
    def __init__(self, n_arms: int):
        #Initialize a confidence-bound agent using the principle of optimism in the face of uncertainty.
        super().__init__(n_arms, "UCB1")

    def select_arm(self, current_step: int) -> int:
        # Initial phase: play each arm exactly once - Cold Start
        if current_step < self.n_arms:
            return current_step
        # UCB Calculation: estimated mean + sqrt((2 * ln(t)) / N_a)
        exploration_factor = np.sqrt((2 * np.log(current_step)) / self.prescription_counts)
        # Every time you pull/prescribe arm a, its N_a/prescription_counts increases, causing its exploration factor to shrink.
        ucb_values = self.estimated_efficacies + exploration_factor
        return int(np.argmax(ucb_values))
```

### Strategy Analysis: Confidence-Based Strategy (UCB1)

**Overview:**
The Upper Confidence Bound (UCB1) algorithm operates on the principle of "optimism in the face of uncertainty." Instead of randomly exploring or relying purely on the current average success rate, it calculates a maximum plausible success rate (the upper confidence bound) for each medicine. The algorithm selects the treatment that maximizes:

$\text{UCB} = \hat{\mu}_a + \sqrt{\frac{2 \ln(t)}{N_a}}$

Where $\hat{\mu}_a$ is the estimated success rate of the medicine, $t$ is the current patient number, and $N_a$ is the number of times this specific medicine has been prescribed.

**Expected Behavior & Trade-offs:**
* **Dynamic, Evidence-Based Exploration:** As requested by the senior physician, the term $\sqrt{\frac{2 \ln(t)}{N_a}}$ artificially boosts the score of medicines that have been tested rarely. As a medicine is tested more frequently ($N_a$ increases), this exploration bonus shrinks, and the algorithm relies heavier on the true observed efficacy ($\hat{\mu}_a$).
* **No Manual Tuning Required:** Unlike Epsilon-Greedy, which requires arbitrarily setting an exploration rate ($\epsilon$), UCB1 naturally and mathematically balances exploration and exploitation on its own.
* **Optimal Real-World Safety:** UCB1 is mathematically proven to minimize long-term regret. After an initial "cold start" phase where it tests every drug, it rapidly zeroes in on the optimal treatment while periodically checking alternatives just enough to be statistically certain. This makes it a highly ethical and effective choice for a real-world clinical trial.

```python
# ==========================================
# MAIN EXECUTION
# ==========================================

# Initialize all requested MAB strategies
strategies = [
    ImmediateExploiter(env.total_medicines),
    EpsilonGreedyStrategy(env.total_medicines, 0.10),
    EpsilonGreedyStrategy(env.total_medicines, 0.01),
    EpsilonGreedyStrategy(env.total_medicines, 0.50),
    UCB1Strategy(env.total_medicines)
]
simulator = TrialSimulation(env)
histories = {}
# Run all strategies sequentially
for strat in strategies:
    hist, _ = simulator.execute(strat)
    histories[strat.name] = hist
```
